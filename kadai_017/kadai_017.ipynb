```python
class Human():
  def __init__(self, name, age):
      self.name=name
      self.age=age

  def check_adult(self):
      if self.age >= 20:
          print("大人である")
      else:
          print("大人でない")

A=Human("太郎",5)
B=Human("二郎",10)
C=Human("三郎",15)
D=Human("史郎",20)
E=Human("五郎",25)
F=Human("六郎",30)
Human_list=[A,B,C,D,E,F]

for Human in Human_list:
    Human.check_adult()



```
